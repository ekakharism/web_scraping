```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

payload = {
    "title": "Belajar Requests",
    "body": "Ini contoh POST request",
    "userId": 1
}

response = requests.post(url, json=payload)

print(response.status_code)
print(response.json())
```

```python
import requests

params = {
    "userId": 1
}

response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params=params
)

print(response.url)  # lihat URL lengkapnya
print(response.json())
```

```python
import requests
import pandas as pd

r = requests.get('https://api.github.com/users/ekaf-pro/repos',
                 params={
                     'sort' : 'updated'
                 })

print(r.status_code)

items = r.json()

type(items)
```

```python
import pandas as pd
r = requests.get('https://api.github.com/search/repositories',
                 params={
                    'q':'stars:>100000',
                    'sort':'stars',
                    'order':'desc',
                    'per_page': 50}
)

print(r.status_code)
items = r.json()['items']
type(items)
```

```python
cols = ['full_name', 'description','html_url','language',
       'stargazers_count','forks_count','open_issues_count',
       'created_at','pushed_at','license','topics']
```

```python
import pandas as pd

df = pd.DataFrame(items)[cols]
df
```

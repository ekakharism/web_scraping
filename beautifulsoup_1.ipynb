```python
import requests # untuk meminta/mengambil kode html dari halaman web
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"
html = requests.get(url).text

soup_quotes = BeautifulSoup(html, 'html.parser')
quotes = soup_quotes.find_all('span', class_="text")

for i, q in enumerate(quotes, start=1):
    print(f'Quote ke-{i}: {q.text}')
```

```python
soup_author = BeautifulSoup(html, 'html.parser')
author = soup_author.find_all('small', class_="author")

for i, q in enumerate(author, start=1):
    print(f'author-{i}: {q.text}')
```

```python
soup_tag = BeautifulSoup(html, 'html.parser')
tag = soup_tag.find_all('div', class_="tags")

for i, q in enumerate(tag, start=1):
    print(f'author-{i}: {q.text}')
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Ambil HTML dari website
url = "https://quotes.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# 2. Temukan semua blok quote
quote_blocks = soup.find_all("div", class_="quote")

# 3. Simpan hasil scraping ke list
data = []

for qb in quote_blocks:
    quote_text = qb.find("span", class_="text").get_text(strip=True)
    author_name = qb.find("small", class_="author").get_text(strip=True)

    data.append({
        "author": author_name,
        "quote": quote_text
    })

# 4. Buat DataFrame
df = pd.DataFrame(data)

df
```

```python
import requests
import pandas as pd 
from bs4 import BeautifulSoup

url = 'https://realpython.github.io/fake-jobs/'
response = requests.get(url)

soup = BeautifulSoup(response.text, 'html.parser')
lowongan_block = soup.find_all('div', class_='media-content')

data = []

for i, q in enumerate(lowongan_block, start=1):
    lowongan = q.find('h2', class_='title').get_text(strip=True)
    perusahaan = q.find('h3', class_='subtitle').get_text(strip=True)
    data.append({
        'Nama Posisi' : lowongan,
        'Perusahaan' : perusahaan
    })

df = pd.DataFrame(data)
df
```
